Tech Stack Requirements Documentation
Project: Staganalysis

1. Overview
This project demonstrates steganography techniques across multiple media formats, including BMP, PNG, TIFF images, MP3/WAV audio, and embedded video files. It provides both command-line scripts and graphical interfaces for encoding, decoding, and recovering hidden data.
The project is modular, with separate folders for each format and a main application (StegApp) to unify the functionality.

2. Programming Language
3.11+

3. Core Dependencies
Image Processing
 - Pillow (PIL) → for image manipulation (BMP, PNG, TIFF).
 - Numpy → efficient array operations for pixel data.
 - OpenCV → could be used for advanced visual analysis.

Audio Processing
 - pydub / wave / struct → for WAV audio read/write.
 - mutagen → MP3 metadata parsing.
 - custom mp3_parser (in repo) → handles low-level MP3 bitstream operations.

GUI / Application
 - Tkinter → lightweight GUI for stego apps (gui.py, ui.py).
 - Custom helper modules → e.g., base_codec.py, image_codec.py, wav_codec.py.

Packaging & Environment
 - Virtual Environment (.venv) included in Stego_Tiff.
 - requirements.txt for dependency installation:
 - pip install -r requirements.txt

4. Tools & Infrastructure
 - Git → Version control (project has .git/ metadata).
 - IDE → Recommended: PyCharm / VS Code for development.
 - Platform → Works on Windows, Linux, Mac (Python-compatible OS).
 - Data → Sample cover files and secret payloads (bmp, png, tiff, wav, mp3, docx, pdf, exe).

5. Modules & Responsibilities
Module	Purpose
StegApp	Main app with codecs for image & audio stego (modular, GUI-supported).
Stego_Bmp	Hides & extracts data in BMP files (text, docx, pdf, exe, images).
Stego_Png	Implements PNG-based steganography with examples.
Stego_Tiff	TIFF stego with GUI and utility scripts.
Stego_mp3	Hides/extracts messages in MP3 audio.
Stego_Wav	WAV steganography, supporting image and video payloads.

6. Example Workflow
 - Choose a module (e.g., Stego_Bmp).
 - Select cover file (cover1.bmp).
 - Run encoder script to hide secret payload (e.g., secret2.docx).
 - Generate stego file (cover1__steg.bmp).
 - Run decoder script to extract hidden payload back into recovered_secret/.
